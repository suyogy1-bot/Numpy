## Q22. Delete rows where Title is missing.

## Q23. Fill missing Rating with mean.

```python
fill_mean = df['Rating'].mean()
df['Rating'] = df['Rating'].fillna(fill_mean)
```

```python
df['Rating'].isnull().sum()
```

```python
df['Rating']
```

## Q24. Fill missing Country with mode.

```python
df['Country'].mode()
```

```python
fill_mode = df['Country'].mode()[0]
df['Country'] = df['Country'].fillna(fill_mode)
```

```python
df['Country'].isnull().sum()
```

```python
df['Country']
```

## Q25. Fill missing Budget with median.

```python
df['Budget'] = pd.to_numeric(df['Budget'], errors='coerce')
```

```python
fill_budget = df['Budget'].median()
df['Budget'] = df['Budget'].fillna(fill_budget)
```

```python
df['Budget']
```
