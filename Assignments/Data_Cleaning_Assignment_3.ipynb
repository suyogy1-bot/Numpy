## 5. Filtering Data (Boolean Conditions)

```python
# Customers older than 40
df_age_gt_40 = df[df['Age'] > 40]

# Orders with Revenue greater than 10,000
df_high_revenue = df[df['Revenue'] > 10000]

# Orders paid by Credit Card
df_cc = df[df['Payment_Method'] == 'Credit Card']
```

```python
# Gold members from Bhopal
df_gold_bhopal = df[
    (df['Membership_Type'] == 'Gold') &
    (df['City'].str.lower() == 'bhopal')
]

# Orders from either Bhopal OR Chennai
df_bhopal_chennai = df[
    (df['City'].str.lower() == 'bhopal') |
    (df['City'].str.lower() == 'chennai')
]

# Non-returned products (not Yes)
df_not_returned = df[df['Returned_Product'] != 'Yes']

# Equivalent using NOT (~)
df_not_returned2 = df[~(df['Returned_Product'] == 'Yes')]
```

```python
# Membership type in a list of categories
df_premium = df[df['Membership_Type'].isin(['Gold', 'Platinum'])]

# Age between 25 and 35 (inclusive)
df_age_25_35 = df[df['Age'].between(25, 35)]

# Rows where City is missing
df_city_null = df[df['City'].isnull()]

# Rows where Profit is not missing
df_profit_not_null = df[df['Profit'].notnull()]
```

```python
# Bhopal + Gold or Platinum + high revenue, show only a few columns
mask = (
    (df['City'].str.lower() == 'bhopal') &
    (df['Membership_Type'].isin(['Gold', 'Platinum'])) &
    (df['Revenue'] > 10000)
)

df_filtered = df.loc[mask,
                     ['Customer_Name', 'City', 'Membership_Type',
                      'Revenue', 'Profit']]
```

## 10. `groupby`, `agg`, and `transform`

```python
df['Profit'] = (
    df['Profit']
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)
```

```python
df['City'].unique()
```

```python
# 1) Normalize to lowercase + strip spaces
city_norm = df['City'].str.strip().str.lower()

# 2) Define mapping on normalized values
city_map = {
    'bhopal': 'Bhopal',
    'jaipur': 'Jaipur',
    'hyderabad': 'Hyderabad',
    'ahmedabad': 'Ahmedabad',
    'indore': 'Indore',
    'lucknow': 'Lucknow',
    'delhi': 'Delhi',
    'new delhi': 'New Delhi',
    'bengaluru': 'Bengaluru',
    'bangalore': 'Bengaluru',
    'pune': 'Pune',
    'mumbai': 'Mumbai',
    'kolkata': 'Kolkata',
    'chennai': 'Chennai',
}

# 3) Apply mapping; keep original where no mapping
df['City'] = city_norm.map(city_map).fillna(df['City'])
```
